```python
import os
import numpy as np
import pickle

data_path = '/data/EEG-X/Processed/TUAB/processed_128hz_1280seqlen/windowed_10S/'
Data = {}
val_data = []
val_data_path = os.path.join(data_path, 'val')
for file in os.listdir(val_data_path):
    if file.endswith('.pkl'):
        with open(os.path.join(val_data_path, file), 'rb') as f:
            val_data.append(pickle.load(f))

# Convert val_data (list of dicts) to arrays for X, X_Clean, y, and id
Val_data = {}
Val_data['X_val'] = np.stack([d['X'] for d in val_data])
Val_data['X_val_clean'] = np.stack([d['X_Clean'] for d in val_data])
Val_data['y_val'] = np.array([d['y'] for d in val_data])
Val_data['id_val'] = np.array([d['id'] for d in val_data])


train_data_path = os.path.join(data_path, 'train')
train_data = []
for file in os.listdir(train_data_path):
    if file.endswith('.pkl'):
        with open(os.path.join(train_data_path, file), 'rb') as f:
            train_data.append(pickle.load(f))

Train_data = {}
Train_data['X_train'] = np.stack([d['X'] for d in train_data])
Train_data['X_train_clean'] = np.stack([d['X_Clean'] for d in train_data])
Train_data['y_train'] = np.array([d['y'] for d in train_data])
Train_data['id_train'] = np.array([d['id'] for d in train_data])


test_data_path = os.path.join(data_path, 'test')
test_data = []
for file in os.listdir(test_data_path):
    if file.endswith('.pkl'):
        with open(os.path.join(test_data_path, file), 'rb') as f:
            test_data.append(pickle.load(f))

Test_data = {}
Test_data['X_test'] = np.stack([d['X'] for d in test_data])
Test_data['X_test_clean'] = np.stack([d['X_Clean'] for d in test_data])
Test_data['y_test'] = np.array([d['y'] for d in test_data])
Test_data['id_test'] = np.array([d['id'] for d in test_data])

print("Train data shape: ", Train_data['X_train'].shape)
print("Val data shape: ", Val_data['X_val'].shape)
print("Test data shape: ", Test_data['X_test'].shape)
```

```python
save_data_dir = '/data/EEG-X/Processed/TUAB/'
 # Save the data using a higher pickle protocol

data_dict = {
    'X_train': Train_data['X_train'], 'Y_train': Train_data['y_train'],
    'X_val': Val_data['X_val'], 'Y_val': Val_data['y_val'], 
    'X_test': Test_data['X_test'], 'Y_test': Test_data['y_test']
}
save_path = save_data_dir + '/TUAB.npy'
with open(save_path, 'wb') as f:
    pickle.dump(data_dict, f)
```
