```python
# Build the dictionary with shuffled data
# Function to apply channel shuffle
import numpy as np
import pickle
import os

Data_npy = np.load('/data/EEG-X/Processed/STEW/STEW.npy', allow_pickle=True)


def shuffle_channels(data):
    return data[:, shuffled_indices, :]

data_dict = {
    'X_train': shuffle_channels(Data_npy['X_train']),
    'Y_train': Data_npy['Y_train'],
    'X_val': shuffle_channels(Data_npy['X_val']),
    'Y_val': Data_npy['Y_val'],
    'X_test': shuffle_channels(Data_npy['X_test']),
    'Y_test': Data_npy['Y_test'],
}

os.makedirs('/data/EEG-X/Processed/STEW_shuffled', exist_ok=True)
# Save the shuffled data dictionary
save_path = '/data/EEG-X/Processed/STEW_shuffled/STEW_shuffled.npy'
with open(save_path, 'wb') as f:
    pickle.dump(data_dict, f)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import expit  # logistic function
from scipy.stats import norm

# Simulated raw outputs from a tanh classifier (range -1 to 1)
np.random.seed(42)
raw_scores = np.tanh(np.random.normal(loc=0.5, scale=1.0, size=1000))

# Step 1: Rescale to (0, 1) using logistic function (squash extremes)
logistic_scaled = expit(raw_scores)

# Step 2: Match CDF to standard normal
# Compute percentiles (empirical CDF)
ranks = np.argsort(np.argsort(logistic_scaled)) / len(logistic_scaled)
normal_mapped = norm.ppf(ranks.clip(1e-5, 1 - 1e-5))  # Avoid infs

# Plotting
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# Raw scores
axs[0].hist(raw_scores, bins=50, color='gray')
axs[0].set_title('Raw Classifier Output (tanh)')
axs[0].set_xlabel('Score')
axs[0].set_ylabel('Count')

# Logistic-scaled scores
axs[1].hist(logistic_scaled, bins=50, color='dodgerblue')
axs[1].set_title('Logistic-scaled Output (0-1)')
axs[1].set_xlabel('Rescaled Score')

# Normal-mapped scores
axs[2].hist(normal_mapped, bins=50, color='darkorange')
axs[2].set_title('Mapped to Standard Normal')
axs[2].set_xlabel('Z-score')

plt.tight_layout()
plt.show()
```
